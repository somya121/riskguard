```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/raw/credit_portfolio.csv"
)

df.head()
```

```python
default_rate = df["default_flag"].mean()

print(
    f"Portfolio default rate: "
    f"{default_rate:.2%}"
)
```

```python
df.groupby(
    pd.qcut(
        df["credit_score"],
        10
    )
)["default_flag"].mean()
```

```python
df.groupby(
    pd.qcut(
        df["debt_to_income"],
        10
    )
)["default_flag"].mean()
```

```python
df.groupby(
    "previous_defaults"
)["default_flag"].mean()
```

```python
df.groupby(
    "previous_defaults"
)["default_flag"].mean()
```

```python
monthly_default = (
    df.groupby(
        "observation_date"
    )["default_flag"]
    .mean()
)
```

```python
monthly_default.plot(
    figsize=(12, 5)
)

plt.title(
    "Monthly Portfolio Default Rate"
)

plt.xlabel("Date")
plt.ylabel("Default Rate")

plt.grid()
plt.show()
```

```python
numeric_columns = [
    "age",
    "annual_income",
    "employment_years",
    "credit_score",
    "debt_to_income",
    "credit_utilization",
    "previous_defaults",
    "delinquencies_12m",
    "loan_amount",
    "interest_rate",
    "unemployment_rate",
    "gdp_growth",
    "default_flag"
]

corr = df[numeric_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title(
    "Credit Portfolio Correlation Matrix"
)

plt.show()
```
